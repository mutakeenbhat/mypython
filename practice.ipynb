```python
print("\nMissing Values:")
print(df.isnull().sum())
```

```python
df['gender'] = df['gender'].map({'male': 0, 'female': 1})
df['bloodtype'] = df['bloodtype'].astype('category').cat.codes
```

```python
from sklearn.model_selection import train_test_split
features=['Bloodcount','age','gender','id','weight','bp','bloodtype']
x=df[features]
y=df['thyriod']
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
from sklearn.ensemble import RandomForestClassifier

model = RandomForestClassifier(n_estimators=100, random_state=42)

model.fit(x_train, y_train)
```

```python
from sklearn.metrics import accuracy_score, classification_report
prediction = model.predict(x_test)
test_acc = accuracy_score(y_test, prediction)
class_report = classification_report(y_test, prediction)
print('test accuracy is: ', test_acc, "\n  classification_report is: \n", class_report)
```
